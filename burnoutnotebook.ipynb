# Step 2: Missing Values Treatment

In this step, we will:
- **Re-examine** the missing values identified in Step 1
- **Categorize** columns by their missing percentage into treatment groups
- **Apply** appropriate imputation or removal strategies for each group
- **Verify** that all missing values have been handled
- **Document & justify** every decision made

> **Note:** Missing values must be treated BEFORE outlier detection, because missing data can bias statistical measures (Q1, Q3, IQR) used for outlier identification.

## 2.1 Categorize Columns by Missing Percentage

We group columns into categories based on how much data is missing, then apply a different strategy for each group:

| Category | Missing % | Strategy | Justification |
|---|---|---|---|
| **No missing** | 0% | No action needed | Data is complete |
| **Low missing** | 0–5% | Fill categorical → **mode**, numeric → **median** | Very few values missing; imputation is reliable |
| **Moderate missing** | 5–50% | Fill categorical → **"Unknown"**, numeric → **median** | Too many to drop rows; "Unknown" preserves data without assuming a category |
| **High missing** | ≥50% | Fill with **"Not Applicable"** | These are conditional columns (e.g., "Which US state?" doesn't apply to non-US respondents) |

```python
# --- Categorize all 97 columns by missing percentage ---
total_rows = len(df)

# Calculate missing % for every column
missing_info = pd.DataFrame({
    'Column': df.columns,
    'Missing_Count': df.isnull().sum().values,
    'Missing_Pct': (df.isnull().sum().values / total_rows * 100).round(2)
})

# Assign categories
def categorize_missing(pct):
    if pct == 0:
        return 'No Missing (0%)'
    elif pct <= 5:
        return 'Low (0-5%)'
    elif pct <= 50:
        return 'Moderate (5-50%)'
    else:
        return 'High (≥50%)'

missing_info['Category'] = missing_info['Missing_Pct'].apply(categorize_missing)

# Print summary by category
print("MISSING VALUES CATEGORIZATION")
print("=" * 70)
for cat in ['No Missing (0%)', 'Low (0-5%)', 'Moderate (5-50%)', 'High (≥50%)']:
    subset = missing_info[missing_info['Category'] == cat]
    print(f"\n{cat}: {len(subset)} columns")
    if cat != 'No Missing (0%)' and len(subset) > 0:
        for _, row in subset.sort_values('Missing_Pct', ascending=False).iterrows():
            col_short = row['Column'][:60] + '...' if len(row['Column']) > 60 else row['Column']
            print(f"  {row['Missing_Pct']:6.2f}%  ({row['Missing_Count']:,} missing)  {col_short}")

print(f"\n{'=' * 70}")
print(f"Total missing cells across entire dataset: {df.isnull().sum().sum():,} / {total_rows * len(df.columns):,}")
print(f"Overall missing rate: {df.isnull().sum().sum() / (total_rows * len(df.columns)) * 100:.2f}%")
```

## 2.2 Apply Missing Values Treatment

Now we apply the strategy defined above to every column with missing values.

```python
# --- Apply missing values treatment based on category ---
print("APPLYING MISSING VALUES TREATMENT")
print("=" * 70)

treatment_counts = {'filled_mode': 0, 'filled_median': 0, 'filled_unknown': 0, 'filled_na': 0}

for _, row in missing_info.iterrows():
    col = row['Column']
    pct = row['Missing_Pct']
    cat = row['Category']
    
    if cat == 'No Missing (0%)':
        continue  # Nothing to do
    
    n_missing = df[col].isnull().sum()
    if n_missing == 0:
        continue  # Double check
    
    # Determine if column is numeric or categorical
    is_numeric = pd.to_numeric(df[col], errors='coerce').notna().sum() > len(df) * 0.5
    
    if cat == 'High (≥50%)':
        # Conditional columns — fill with "Not Applicable"
        df[col] = df[col].fillna('Not Applicable')
        treatment_counts['filled_na'] += 1
        print(f"  HIGH   ({pct:5.1f}%) {col[:55]:55s} → filled {n_missing:,} with 'Not Applicable'")
    
    elif cat == 'Moderate (5-50%)':
        if is_numeric:
            median_val = pd.to_numeric(df[col], errors='coerce').median()
            df[col] = df[col].fillna(median_val)
            treatment_counts['filled_median'] += 1
            print(f"  MOD    ({pct:5.1f}%) {col[:55]:55s} → filled {n_missing:,} with median ({median_val:.1f})")
        else:
            df[col] = df[col].fillna('Unknown')
            treatment_counts['filled_unknown'] += 1
            print(f"  MOD    ({pct:5.1f}%) {col[:55]:55s} → filled {n_missing:,} with 'Unknown'")
    
    elif cat == 'Low (0-5%)':
        if is_numeric:
            median_val = pd.to_numeric(df[col], errors='coerce').median()
            df[col] = df[col].fillna(median_val)
            treatment_counts['filled_median'] += 1
            print(f"  LOW    ({pct:5.1f}%) {col[:55]:55s} → filled {n_missing:,} with median ({median_val:.1f})")
        else:
            mode_val = df[col].mode()[0] if len(df[col].mode()) > 0 else 'Unknown'
            df[col] = df[col].fillna(mode_val)
            treatment_counts['filled_mode'] += 1
            print(f"  LOW    ({pct:5.1f}%) {col[:55]:55s} → filled {n_missing:,} with mode ('{mode_val[:20]}')")

print(f"\n{'=' * 70}")
print(f"TREATMENT SUMMARY:")
print(f"  Filled with mode (low missing, categorical):    {treatment_counts['filled_mode']} columns")
print(f"  Filled with median (low/moderate, numeric):     {treatment_counts['filled_median']} columns")
print(f"  Filled with 'Unknown' (moderate, categorical):  {treatment_counts['filled_unknown']} columns")
print(f"  Filled with 'Not Applicable' (high, conditional): {treatment_counts['filled_na']} columns")
```

## 2.3 Verify — Confirm No Missing Values Remain

After treatment, we check that the dataset has zero missing values.

```python
# --- Verify: check for any remaining missing values ---
remaining_missing = df.isnull().sum()
remaining_total = remaining_missing.sum()

print("VERIFICATION — MISSING VALUES AFTER TREATMENT")
print("=" * 70)
print(f"Total missing values remaining: {remaining_total:,}")
print(f"Dataset shape: {df.shape[0]:,} rows × {df.shape[1]} columns")

if remaining_total == 0:
    print("\n✓ SUCCESS: All missing values have been handled. The dataset is complete.")
else:
    print(f"\n✗ WARNING: {remaining_total:,} missing values still remain in {(remaining_missing > 0).sum()} columns:")
    for col in remaining_missing[remaining_missing > 0].index:
        print(f"  {col[:60]:60s}  → {remaining_missing[col]:,} missing")

# Show a quick before/after comparison
print(f"\n{'=' * 70}")
print("BEFORE vs AFTER comparison:")
print(f"  Before treatment: {missing_info['Missing_Count'].sum():,} missing cells")
print(f"  After treatment:  {remaining_total:,} missing cells")
print(f"  Rows preserved:   {len(df):,} (no rows were dropped)")
```

## 2.5 Step 2 Summary — Missing Values Treatment

### What was done:

| Category | Missing % | # Columns | Strategy Applied |
|---|---|---|---|
| **No missing** | 0% | — | No action |
| **Low missing** | 0–5% | varies | Categorical → **mode** (most frequent value); Numeric → **median** |
| **Moderate missing** | 5–50% | varies | Categorical → **"Unknown"**; Numeric → **median** |
| **High missing** | ≥50% | varies | All → **"Not Applicable"** (conditional questions that didn't apply) |

### Justification:
- **Mode imputation (low missing, categorical):** With very few values missing (<5%), the most frequent category is the best single-value estimate. The impact on distributions is minimal.
- **Median imputation (numeric columns):** Median is robust to outliers (which we haven't treated yet) and preserves the center of the distribution.
- **"Unknown" (moderate missing, categorical):** When 5–50% is missing, imputing with the mode would artificially inflate one category. "Unknown" is honest — it preserves all rows while clearly flagging incomplete data.
- **"Not Applicable" (high missing, conditional):** Columns with ≥50% missing are conditional questions (e.g., "Which US state do you live in?" only applies to US respondents). These are not truly "missing" — the question simply didn't apply. "Not Applicable" correctly reflects this.
- **No rows were dropped** — all 20,959 responses are preserved.

Most of the 97 columns are categorical (text-based survey responses). We select only the columns that contain **continuous numeric values** where outliers are meaningful. Some were originally stored as text (due to commas or mixed entries), so we convert them to numeric first.

```python
# --- Identify the columns that should be numeric for outlier analysis ---
# These are the meaningful continuous/numeric survey columns:
outlier_candidates = {
    'Remote_Hours': 'On.average..what.amount.of.time.of.your.weekly.work.schedule.do.you.perform.remotely..in.the.office.......Remotely...Hours.',
    'Office_Hours': 'On.average..what.amount.of.time.of.your.weekly.work.schedule.do.you.perform.remotely..in.the.office.......In.the.Office..Hours.',
    'Commute_Minutes': 'How.much.time.do.you.spend.on.commuting.to.work..in.minutes.per.commute..',
    'Age': 'Please.share.the.following...Your.age.in.years..',
    'Education_Years': 'Please.share.the.following...Total.years.of.full.time.education..from.primary.school.to.higher.education...',
    'Work_Experience': 'Please.share.the.following...Total.years.of.professional.working.life..count.part.time.work.as.a.fraction..e.g...0.5.for.half.time.work...'
}

# Convert to numeric (coerce errors to NaN — handles commas, text entries, "Not Applicable" from Step 2)
for short_name, full_name in outlier_candidates.items():
    df[full_name] = df[full_name].astype(str).str.replace(',', '.', regex=False)
    df[full_name] = pd.to_numeric(df[full_name], errors='coerce')

# Show basic stats for these columns BEFORE any outlier treatment
print("NUMERIC COLUMNS FOR OUTLIER ANALYSIS (after missing values treatment, before outlier treatment)")
print("=" * 90)
for short_name, full_name in outlier_candidates.items():
    col_data = df[full_name].dropna()
    n_nan = df[full_name].isnull().sum()
    print(f"\n{short_name} ({full_name[:60]}...)")
    print(f"  Count: {len(col_data):,}  |  NaN (non-numeric entries): {n_nan:,}")
    print(f"  Min: {col_data.min():.1f}  |  Max: {col_data.max():.1f}")
    print(f"  Mean: {col_data.mean():.2f}  |  Median: {col_data.median():.1f}")
    print(f"  Std: {col_data.std():.2f}")
```

## 3.3 Outlier Detection — IQR Method & Domain Knowledge

We use **two strategies** combined:
1. **Domain-based bounds** — logical limits based on real-world constraints (e.g., a work week cannot exceed 168 hours, age must be 18–100)
2. **IQR method** — flag statistical outliers beyond Q1 − 1.5×IQR and Q3 + 1.5×IQR

Values outside domain bounds → **removed (set to NaN)** as they are clearly data entry errors.  
Values that are plausible but statistically extreme → **capped (winsorized)** to preserve the data.

```python
# --- Define domain-based bounds for each column ---
# These are hard limits: values outside are IMPOSSIBLE and must be data entry errors
domain_bounds = {
    'Remote_Hours':    (0, 168),    # Max hours in a week
    'Office_Hours':    (0, 168),    # Max hours in a week
    'Commute_Minutes': (0, 300),    # 5-hour one-way commute is extreme but possible
    'Age':             (18, 100),   # Survey requires 18+; 100 is a reasonable upper bound
    'Education_Years': (0, 30),     # Max ~30 years from primary to PhD
    'Work_Experience': (0, 60),     # 60 years of work experience is the maximum realistic
}

# --- Detect outliers: domain violations + IQR ---
print("OUTLIER DETECTION REPORT (on cleaned data — post missing values treatment)")
print("=" * 90)

for short_name, full_name in outlier_candidates.items():
    col_data = df[full_name]
    lo, hi = domain_bounds[short_name]
    
    # Count domain violations (impossible values)
    domain_violations = ((col_data < lo) | (col_data > hi)) & col_data.notna()
    n_domain = domain_violations.sum()
    
    # IQR on valid-domain data only
    valid = col_data[(col_data >= lo) & (col_data <= hi)]
    Q1 = valid.quantile(0.25)
    Q3 = valid.quantile(0.75)
    IQR = Q3 - Q1
    iqr_lo = Q1 - 1.5 * IQR
    iqr_hi = Q3 + 1.5 * IQR
    iqr_outliers = ((valid < iqr_lo) | (valid > iqr_hi)).sum()
    
    print(f"\n{'─' * 90}")
    print(f"  {short_name}")
    print(f"  Domain bounds: [{lo}, {hi}]")
    print(f"  Domain violations (REMOVE → NaN): {n_domain:,}")
    print(f"  IQR bounds: [{iqr_lo:.1f}, {iqr_hi:.1f}]")
    print(f"  IQR outliers (CAP to bounds):     {iqr_outliers:,}")
    
    # Show some example bad values
    bad_vals = col_data[domain_violations].head(5).tolist()
    if bad_vals:
        print(f"  Example impossible values: {bad_vals}")
```

## 3.4 Apply Outlier Treatment

**Strategy per column:**

| Column | Domain Violations | Action | IQR Outliers | Action |
|---|---|---|---|---|
| Remote_Hours | 7 values > 168h | **Remove** (set NaN) — impossible | 260 | **Cap** to IQR bounds |
| Office_Hours | 5 values > 168h | **Remove** (set NaN) — impossible | 135 | **Cap** to IQR bounds |
| Commute_Minutes | 79 values > 300min | **Remove** (set NaN) — 5h+ one-way is data error | 1,164 | **Cap** to IQR bounds |
| Age | 55 values (years like 1998, or 0) | **Remove** (set NaN) — entered birth year instead of age | 629 | **Cap** to IQR bounds |
| Education_Years | 23 values > 30 | **Remove** (set NaN) — entered calendar year | 1,443 | **Cap** to IQR bounds |
| Work_Experience | 7 values > 60 | **Remove** (set NaN) — impossible experience | 647 | **Cap** to IQR bounds |

**Justification:**
- **Domain violations → REMOVED** because they are clearly data entry errors (typed a date, added extra digits, etc.)
- **IQR outliers → CAPPED (winsorized)** because they are plausible but extreme; capping preserves data while reducing skew

```python
# --- Apply outlier treatment: Step 1 = Remove domain violations, Step 2 = Cap IQR outliers ---
print("APPLYING OUTLIER TREATMENT")
print("=" * 80)

treatment_log = []  # Track all changes for documentation

for short_name, full_name in outlier_candidates.items():
    lo_domain, hi_domain = domain_bounds[short_name]
    
    # --- Step 1: Remove domain violations (set to NaN) ---
    mask_invalid = (df[full_name] < lo_domain) | (df[full_name] > hi_domain)
    n_removed = mask_invalid.sum()
    df.loc[mask_invalid, full_name] = np.nan
    
    # --- Step 2: Cap remaining values using IQR ---
    valid = df[full_name].dropna()
    Q1 = valid.quantile(0.25)
    Q3 = valid.quantile(0.75)
    IQR = Q3 - Q1
    iqr_lo = max(lo_domain, Q1 - 1.5 * IQR)  # Don't cap below domain minimum
    iqr_hi = Q3 + 1.5 * IQR
    
    n_capped_lo = (df[full_name] < iqr_lo).sum()
    n_capped_hi = (df[full_name] > iqr_hi).sum()
    
    df[full_name] = df[full_name].clip(lower=iqr_lo, upper=iqr_hi)
    
    # Log the treatment
    treatment_log.append({
        'Column': short_name,
        'Domain Removed': n_removed,
        'Capped Low': n_capped_lo,
        'Capped High': n_capped_hi,
        'IQR Lower': round(iqr_lo, 1),
        'IQR Upper': round(iqr_hi, 1)
    })
    
    print(f"\n  {short_name}:")
    print(f"    ✗ Removed {n_removed:,} impossible values → NaN")
    print(f"    ↓ Capped {n_capped_lo:,} low values to {iqr_lo:.1f}")
    print(f"    ↑ Capped {n_capped_hi:,} high values to {iqr_hi:.1f}")

# Summary table
print(f"\n{'=' * 80}")
print("\nTREATMENT SUMMARY TABLE:")
treatment_df = pd.DataFrame(treatment_log)
print(treatment_df.to_string(index=False))
```

# Step 5: Feature Scaling

In this step, we will normalize/standardize numerical features so all columns are on a consistent scale.

**Why scale?**
- Many ML algorithms (KNN, SVM, logistic regression, PCA, neural networks) are **distance-based** — features with larger ranges dominate the model.
- Example: `Commute_Minutes` (0–94) vs. a binary dummy column (0–1) → without scaling, commute minutes has ~94× more influence.

**Strategy:**
| Column Type | Scaling Method | Reason |
|---|---|---|
| **Continuous numeric** (hours, minutes, age, education, experience) | **StandardScaler** (z-score: mean=0, std=1) | Handles the varying ranges; robust for most ML algorithms |
| **Ordinal Likert** (1–7) | **StandardScaler** | Same bounded range, but scaling ensures consistency with other features |
| **Binary & One-Hot** (0/1) | **StandardScaler** applied uniformly | Keeps everything on the same scale; minor effect on 0/1 columns but ensures consistency |
| **Label-encoded nominal** (0–N integers) | **StandardScaler** | These are arbitrary integers; scaling prevents them from dominating |

> **Note:** We apply StandardScaler to ALL columns uniformly for simplicity and consistency. The result is stored in `df_scaled`.

## 5.2 Apply StandardScaler (Z-Score Normalization)

StandardScaler transforms each feature to: $z = \frac{x - \mu}{\sigma}$

After scaling, every column will have **mean ≈ 0** and **std ≈ 1**.

```python
# --- Apply StandardScaler to all columns ---
scaler = StandardScaler()

# Handle NaN values: StandardScaler doesn't handle NaN → fill with column mean first
# (Only 176 NaN from outlier treatment's domain violations in 6 numeric columns)
nan_count_before = df_encoded.isnull().sum().sum()
print(f"NaN values before scaling: {nan_count_before}")

if nan_count_before > 0:
    # Fill NaN with column mean (standard approach before scaling)
    df_encoded = df_encoded.fillna(df_encoded.mean())
    print(f"Filled NaN with column means. NaN remaining: {df_encoded.isnull().sum().sum()}")

# Fit and transform
df_scaled = pd.DataFrame(
    scaler.fit_transform(df_encoded),
    columns=df_encoded.columns,
    index=df_encoded.index
)

print(f"\n✓ StandardScaler applied to all {df_scaled.shape[1]} columns.")
print(f"  Shape: {df_scaled.shape[0]:,} rows × {df_scaled.shape[1]} columns")
print(f"  All values are now z-scores (mean≈0, std≈1)")
```

# Step 6: Target Engineering & Train/Test Split

Since the dataset contains no direct "Burnout Risk" column, we **engineer a composite Burnout Risk Score** from the three most direct burnout predictors:

| Column | Role | Direction |
|--------|------|-----------|
| #8 - Remote work positive personal impact | Direct predictor | Inverse (higher = less burnout) |
| #11 - Remote work negative career impact | Direct predictor | Direct (higher = more burnout) |
| #12 - Work-life balance difficulty (1–7) | Direct predictor | Direct (higher = more burnout) |

**Composite formula:**  
`Burnout_Risk_Score = (#11 + #12) - #8`  
Thresholded at the **median** into a binary label:

| Label | Value | Condition |
|-------|-------|-----------|
| 🟢 Low Risk | 0 | score < median |
| 🔴 High Risk | 1 | score ≥ median |

- **Target variable (y):** Burnout Risk (binary: 0 = Low Risk, 1 = High Risk)
- **Features (X):** All remaining columns (excluding #8, #11, #12 used to build the target)
- **Split ratio:** 80% train / 20% test
- **Random state:** 42 (for reproducibility)


# 6.1 Engineer Target (y) and Define Features (X)

We engineer the **Burnout Risk** 3-class target variable and exclude the source columns from features to avoid data leakage:
- **#8, #11, #12** are used to build the target — keeping them as features would let the model "cheat"
- Thresholds: 33rd percentile (Low→Medium boundary) and 66th percentile (Medium→High boundary)


```python
# --- Engineer Burnout Risk target variable and define features ---
from sklearn.model_selection import train_test_split
import numpy as np

# Identify the 8 Likert columns (same order as original df)
likert_cols = [c for c in df_scaled.columns if c.startswith('Please.rate')]

# Burnout-relevant columns (by position in Likert block)
col_positive_impact  = likert_cols[1]   # #8  - Remote work positively impacts personal life (INVERSE)
col_negative_career  = likert_cols[4]   # #11 - Remote work negatively impacts career (DIRECT)
col_worklife_balance = likert_cols[5]   # #12 - Difficult to maintain work-life balance (DIRECT)

print("BURNOUT TARGET COLUMNS CHECK")
print("=" * 80)
print(f"[+] Positive personal impact : {col_positive_impact[:80]}")
print(f"[+] Negative career impact   : {col_negative_career[:80]}")
print(f"[+] Work-life balance diff.  : {col_worklife_balance[:80]}")

# --- Engineer composite Burnout Risk Score ---
burnout_score = (
    df_scaled[col_negative_career] +
    df_scaled[col_worklife_balance] -
    df_scaled[col_positive_impact]
)

# Binary label using median threshold
median_threshold = burnout_score.median()
y = (burnout_score >= median_threshold).astype(int)

print(f"\nBURNOUT RISK SCORE STATS")
print(f"  Median threshold : {median_threshold:.4f}")
print(f"  Low Risk  (0)    : {(y==0).sum():,} ({(y==0).mean()*100:.1f}%)")
print(f"  High Risk (1)    : {(y==1).sum():,} ({(y==1).mean()*100:.1f}%)")

# --- Define X: drop the 3 columns used to build the target ---
leakage_cols = [col_positive_impact, col_negative_career, col_worklife_balance]
X = df_scaled.drop(columns=leakage_cols)

print(f"\nFEATURES (X):")
print(f"  Shape  : {X.shape[0]:,} rows x {X.shape[1]} columns")
print(f"  Dropped: {len(leakage_cols)} burnout source columns (leakage prevention)")

print(f"\nTARGET (y):")
print(f"  Shape  : {y.shape[0]:,} values")
print(f"  Type   : Binary classification (0 = Low Risk, 1 = High Risk)")
```

## 6.2 Split into Train and Test Sets

We use an **80/20 split** with `random_state=42` for reproducibility. Since this is a regression task, a simple random split is appropriate (no stratification needed).

```python
# --- 80/20 Train/Test Split ---
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y        # ensures equal Low/High ratio in train and test
)

# Print split results
print("TRAIN / TEST SPLIT RESULTS")
print("=" * 80)
print(f"\nTRAINING SET:")
print(f"  X_train shape: {X_train.shape[0]:,} rows x {X_train.shape[1]} columns")
print(f"  y_train — Low Risk  (0): {(y_train==0).sum():,} ({(y_train==0).mean()*100:.1f}%)")
print(f"  y_train — High Risk (1): {(y_train==1).sum():,} ({(y_train==1).mean()*100:.1f}%)")

print(f"\nTEST SET:")
print(f"  X_test shape:  {X_test.shape[0]:,} rows x {X_test.shape[1]} columns")
print(f"  y_test  — Low Risk  (0): {(y_test==0).sum():,} ({(y_test==0).mean()*100:.1f}%)")
print(f"  y_test  — High Risk (1): {(y_test==1).sum():,} ({(y_test==1).mean()*100:.1f}%)")

train_pct = X_train.shape[0] / (X_train.shape[0] + X_test.shape[0]) * 100
test_pct  = X_test.shape[0]  / (X_train.shape[0] + X_test.shape[0]) * 100
print(f"\nPROPORTIONS:")
print(f"  Train: {train_pct:.1f}% ({X_train.shape[0]:,} samples)")
print(f"  Test:  {test_pct:.1f}% ({X_test.shape[0]:,} samples)")

overlap = set(X_train.index) & set(X_test.index)
print(f"\n  Index overlap: {len(overlap)} (should be 0)")
print(f"  Total samples: {X_train.shape[0] + X_test.shape[0]:,} (original: {X.shape[0]:,})")
```

# Step 7: Feature Creation (Feature Engineering)

We derive **new features** from existing columns to capture non-linear relationships and interactions that Linear Regression cannot learn on its own. This is done **after the train/test split** to prevent data leakage.

**New features to create:**
1. **Polynomial features** — squared terms for continuous variables (captures non-linear effects)
2. **Interaction terms** — products of key variable pairs (captures combined effects)
3. **Attitude composite** — average of all 8 Likert attitude columns (overall remote work sentiment score)

All transformations are applied **identically but independently** to train and test sets.

## 7.1 Create New Features

We apply the **same transformations** to both train and test sets independently using a reusable function. This ensures no information leaks from test to train.

```python
# --- Feature Engineering Function ---
# Applied identically to train and test to prevent data leakage

def create_features(df_input):
    df_fe = df_input.copy()
    
    commute_col = 'How.much.time.do.you.spend.on.commuting.to.work..in.minutes.per.commute..'
    age_col     = 'Please.share.the.following...Your.age.in.years..'
    edu_col     = 'Please.share.the.following...Total.years.of.full.time.education..from.primary.school.to.higher.education...'
    exp_col     = 'Please.share.the.following...Total.years.of.professional.working.life..count.part.time.work.as.a.fraction..e.g...0.5.for.half.time.work...'
    
    # --- 1. POLYNOMIAL FEATURES ---
    df_fe['Commute_squared']    = df_fe[commute_col] ** 2
    df_fe['Age_squared']        = df_fe[age_col] ** 2
    df_fe['Education_squared']  = df_fe[edu_col] ** 2
    df_fe['Experience_squared'] = df_fe[exp_col] ** 2
    
    # --- 2. INTERACTION TERMS ---
    df_fe['Age_x_Commute']          = df_fe[age_col] * df_fe[commute_col]
    df_fe['Experience_x_Education'] = df_fe[exp_col] * df_fe[edu_col]
    df_fe['Age_x_Experience']       = df_fe[age_col] * df_fe[exp_col]
    df_fe['Commute_x_Education']    = df_fe[commute_col] * df_fe[edu_col]
    
    # --- 3. REMAINING ATTITUDE COMPOSITE (5 Likert cols) ---
    likert_cols_fe = [c for c in df_fe.columns if c.startswith('Please.rate')]
    df_fe['Attitude_composite'] = df_fe[likert_cols_fe].mean(axis=1)
    
    return df_fe

# Apply to BOTH train and test independently
print("FEATURE CREATION")
print("=" * 80)

cols_before = X_train.shape[1]
X_train_fe  = create_features(X_train)
X_test_fe   = create_features(X_test)

cols_after = X_train_fe.shape[1]
new_cols   = cols_after - cols_before

print(f"\n  Features before    : {cols_before}")
print(f"  Features after     : {cols_after}")
print(f"  New features added : {new_cols}")

new_feature_names = [c for c in X_train_fe.columns if c not in X_train.columns]
print(f"\n  NEW FEATURES:")
for i, f in enumerate(new_feature_names, 1):
    print(f"    {i:2d}. {f}")

assert list(X_train_fe.columns) == list(X_test_fe.columns), "Column mismatch!"
print(f"\n  X_train_fe: {X_train_fe.shape[0]:,} x {X_train_fe.shape[1]}")
print(f"  X_test_fe : {X_test_fe.shape[0]:,} x {X_test_fe.shape[1]}")

X_train = X_train_fe
X_test  = X_test_fe
```

# Step 9: Remove Multicollinearity

Highly correlated features carry **redundant information**, inflate variance of regression coefficients, and can cause **overfitting**. We eliminate them systematically:

**Strategy:**
1. Compute the full pairwise Pearson correlation matrix on `X_train`
2. For every pair with |r| > **0.85**, drop the feature that has the **lower absolute correlation** with the target (keeps the more informative one)
3. Apply the same column drop to `X_test` so both sets remain aligned
4. Visualise before / after heatmaps to confirm reduction

```python
# --- Step 9: Multicollinearity Removal ---
import numpy as np

threshold = 0.85

corr_matrix = X_train.corr().abs()
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

# Correlation of each feature with target
target_corr = X_train.corrwith(y_train).abs()

# Find columns to drop (keep the one more correlated with burnout risk)
cols_to_drop = set()
for col in upper.columns:
    correlated_features = upper.index[upper[col] > threshold].tolist()
    for feat in correlated_features:
        if col not in cols_to_drop and feat not in cols_to_drop:
            # Drop the one with lower correlation to target
            if target_corr.get(col, 0) >= target_corr.get(feat, 0):
                cols_to_drop.add(feat)
            else:
                cols_to_drop.add(col)

cols_to_drop = list(cols_to_drop)

print("MULTICOLLINEARITY REMOVAL")
print("=" * 60)
print(f"  Threshold          : |r| > {threshold}")
print(f"  Features before    : {X_train.shape[1]}")
print(f"  Features dropped   : {len(cols_to_drop)}")

# Apply to both train and test
X_train = X_train.drop(columns=cols_to_drop)
X_test  = X_test.drop(columns=cols_to_drop)

print(f"  Features after     : {X_train.shape[1]}")
print(f"\n  X_train: {X_train.shape[0]:,} x {X_train.shape[1]}")
print(f"  X_test : {X_test.shape[0]:,} x {X_test.shape[1]}")

print(f"\nDROPPED COLUMNS:")
for i, c in enumerate(cols_to_drop, 1):
    print(f"  {i:2d}. {c[:70]}")
```

### 9.2 — Verification: Remaining Multicollinearity

## Step 10: Feature Selection

After reducing multicollinearity, we identify the most relevant features to improve performance and interpretability.

### 🔹 Strategy: Ensemble of 3 Methods
We rank features using:
1. **Pearson Correlation** → captures linear relationships with the target  
2. **Mutual Information** → captures linear + non-linear dependencies  
3. **Lasso Regression (L1)** → embedded method that shrinks irrelevant features to zero  

➡️ A **consensus score** is computed by averaging feature ranks across the 3 methods.  
➡️ Features consistently ranked high are considered more important.

---


### 10.1 — Method 1: Pearson Correlation Ranking

```python
# ============================================================
# 10.1 — Method 1: Pearson Correlation with Burnout Risk
# ============================================================
import numpy as np
import pandas as pd

# Compute absolute Pearson correlation of each feature with y_train
pearson_scores = X_train.corrwith(y_train).abs().sort_values(ascending=False)

# Rank features (rank 1 = most correlated)
pearson_rank = pearson_scores.rank(ascending=False).astype(int)

print("METHOD 1: Pearson Correlation Ranking")
print("=" * 60)
print(f"Total features: {len(pearson_scores)}")
print(f"\nTop 20 features by |Pearson r| with Burnout Risk:")
print(f"{'Rank':<6} {'|r|':<8} {'Feature'}")
print("-" * 60)
for i, (feat, score) in enumerate(pearson_scores.head(20).items(), 1):
    f_short = (feat[:50] + '...') if len(feat) > 50 else feat
    print(f"{i:<6} {score:<8.4f} {f_short}")
```

### 10.2 — Method 2: Mutual Information Regression

```python
# ============================================================
# 10.2 — Method 2: Mutual Information (non-linear + linear)
# ============================================================
from sklearn.feature_selection import mutual_info_classif

# Compute MI scores for classification (higher = more informative)
mi_scores = mutual_info_classif(X_train, y_train, random_state=42, n_neighbors=5)
mi_series = pd.Series(mi_scores, index=X_train.columns).sort_values(ascending=False)

# Rank features (rank 1 = highest MI)
mi_rank = mi_series.rank(ascending=False).astype(int)

print("METHOD 2: Mutual Information Classification Ranking")
print("=" * 60)
print(f"Total features: {len(mi_series)}")
print(f"\nTop 20 features by Mutual Information with Burnout Risk:")
print(f"{'Rank':<6} {'MI':<8} {'Feature'}")
print("-" * 60)
for i, (feat, score) in enumerate(mi_series.head(20).items(), 1):
    f_short = (feat[:50] + '...') if len(feat) > 50 else feat
    print(f"{i:<6} {score:<8.4f} {f_short}")
```

### 10.3 — Method 3: Lasso Regression (L1 Regularization)

```python
# ============================================================
# 10.3 — Method 3: L1 Logistic Regression feature importance
# ============================================================
from sklearn.linear_model import LogisticRegression

# L1 Logistic Regression for multiclass classification
lasso = LogisticRegression(penalty='l1', solver='saga', C=0.1,
                           max_iter=1000, random_state=42, n_jobs=-1)
lasso.fit(X_train, y_train)

# Average absolute coefficient across the 3 classes
lasso_coefs = pd.Series(
    np.abs(lasso.coef_).mean(axis=0),
    index=X_train.columns
).sort_values(ascending=False)

# Rank features (rank 1 = largest absolute coefficient)
lasso_rank = lasso_coefs.rank(ascending=False).astype(int)

# Count features with zero coefficient (eliminated by L1)
n_zero    = (lasso_coefs == 0).sum()
n_nonzero = (lasso_coefs > 0).sum()

print("METHOD 3: L1 Logistic Regression Ranking")
print("=" * 60)
print(f"Features with non-zero coefficient: {n_nonzero}")
print(f"Features zeroed out (eliminated):   {n_zero}")
print(f"\nTop 20 features by |L1 coefficient| with Burnout Risk:")
print(f"{'Rank':<6} {'|Coef|':<10} {'Feature'}")
print("-" * 60)
for i, (feat, score) in enumerate(lasso_coefs.head(20).items(), 1):
    f_short = (feat[:48] + '...') if len(feat) > 48 else feat
    print(f"{i:<6} {score:<10.4f} {f_short}")
```

### 10.4 — Consensus Ranking 

We combine the three methods by computing the **average rank** across all three. Features that rank highly in multiple methods are the most robust choices.

```python
# ============================================================
# 10.4 — Consensus Ranking (average rank across 3 methods)
# ============================================================

# Build ranking DataFrame
ranking_df = pd.DataFrame({
    'Pearson_score': pearson_scores,
    'Pearson_rank':  pearson_rank,
    'MI_score':      mi_series,
    'MI_rank':       mi_rank,
    'Lasso_score':   lasso_coefs,
    'Lasso_rank':    lasso_rank
})

# Average rank (lower = better)
ranking_df['Avg_rank'] = ranking_df[['Pearson_rank', 'MI_rank', 'Lasso_rank']].mean(axis=1)
ranking_df = ranking_df.sort_values('Avg_rank')

# Features considered "good" by each method
good_pearson = (pearson_scores > 0.01).sum()
good_mi      = (mi_series > 0).sum()
good_lasso   = (lasso_coefs > 0).sum()

print("CONSENSUS RANKING — Burnout Risk Feature Selection")
print("=" * 90)
print(f"\nFEATURES WITH NON-ZERO / MEANINGFUL SCORES:")
print(f"  Pearson  |r| > 0.01  : {good_pearson} features")
print(f"  MI       > 0         : {good_mi} features")
print(f"  L1 Lasso > 0         : {good_lasso} features (non-zeroed by L1)")
print(f"\n  → Recommended selection: top {good_lasso} features (L1 non-zero as guide)")

print(f"\n\n{'#':<4} {'Avg Rank':>9} {'Pearson R':>9} {'MI Rank':>9} {'Lasso R':>9}  {'Feature'}")
print("-" * 90)
for i, (feat, row) in enumerate(ranking_df.iterrows(), 1):
    f_short = (feat[:45] + '...') if len(feat) > 45 else feat
    print(f"{i:<4} {row['Avg_rank']:9.1f} {row['Pearson_rank']:9.0f} {row['MI_rank']:9.0f} {row['Lasso_rank']:9.0f}  {f_short}")
```


## Step 11: Feature Selection via Cross-Validation

Instead of fixing the number of features manually, we determine the optimal count using **Random Forest + 5-fold cross-validation**.

### 🔹 Approach
1. Rank all 159 features using the consensus score  
2. Evaluate different feature subset sizes:  
   `5, 10, 15, 20, 30, 40, 50, 60, 80, 100`  
3. For each subset:
   - Train a Random Forest model  
   - Perform 5-fold cross-validation  
   - Record mean performance (accuracy / R²)  
4. Select the optimal number of features at the **elbow point**  
   (where performance improvement plateaus)

---

### ✅ Outcome
A robust subset of features that:
- Maximizes predictive performance  
- Reduces overfitting  
- Improves interpretability  

```python
# ============================================================
# 11 — Cross-Validation to find optimal feature count
# ============================================================
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score
import time

feature_counts = [5, 10, 15, 20, 30, 40, 50, 60, 80, 100]
cv_scores = []

rf = RandomForestClassifier(n_estimators=50, max_depth=10,
                            random_state=42, n_jobs=-1)

print("CROSS-VALIDATION — Optimal Feature Count for Burnout Risk")
print("=" * 60)
for n in feature_counts:
    top_cols = ranking_df.head(n).index.tolist()
    start = time.time()
    scores = cross_val_score(rf, X_train[top_cols], y_train,
                             cv=3, scoring='accuracy', n_jobs=-1)
    elapsed = time.time() - start
    mean_acc = scores.mean()
    cv_scores.append(mean_acc)
    print(f"  TOP {n:3d} features → accuracy = {mean_acc:.4f} ± {scores.std():.4f}  ({elapsed:.1f}s)")

best_n   = feature_counts[np.argmax(cv_scores)]
best_acc = max(cv_scores)

print(f"\nBEST FEATURE COUNT : {best_n}")
print(f"BEST CV ACCURACY   : {best_acc:.4f}")

# Apply final selection
selected_features = ranking_df.head(best_n).index.tolist()
X_train = X_train[selected_features]
X_test  = X_test[selected_features]

print(f"\nX_train: {X_train.shape[0]:,} x {X_train.shape[1]}")
print(f"X_test : {X_test.shape[0]:,} x {X_test.shape[1]}")
```

```python
# ============================================================
# 12.1 — Random Forest with Optimized Hyperparameters (15 features)
# ============================================================
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import numpy as np
import time

param_grid = {
    'n_estimators':      [200, 300, 500, 700, 1000],
    'max_depth':         [10, 20, 30, 40, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf':  [1, 2, 4],
    'max_features':      ['sqrt', 'log2', 0.3, 0.5],
    'bootstrap':         [True, False],
    'class_weight':      ['balanced', 'balanced_subsample', None]
}

rf_base = RandomForestClassifier(random_state=42, n_jobs=-1)

rf_search = RandomizedSearchCV(
    rf_base,
    param_distributions=param_grid,
    n_iter=30,
    cv=3,
    scoring='accuracy',
    random_state=42,
    n_jobs=-1,
    verbose=1
)

print("Training Random Forest (15 features)...")
print("=" * 60)
print(f"  X_train: {X_train.shape[0]:,} x {X_train.shape[1]}")
start = time.time()
rf_search.fit(X_train, y_train)
elapsed = time.time() - start

print(f"\nSearch completed in {elapsed:.1f}s")
print(f"Best parameters:")
for k, v in rf_search.best_params_.items():
    print(f"  {k:<25s} : {v}")
print(f"\nBest CV Accuracy : {rf_search.best_score_:.4f}")

# --- Evaluate on test set ---
best_rf = rf_search.best_estimator_
y_pred_rf = best_rf.predict(X_test)
acc_rf = accuracy_score(y_test, y_pred_rf)

print(f"\nTEST SET RESULTS")
print("=" * 60)
print(f"Accuracy : {acc_rf:.4f}")
print(f"\nClassification Report:")
print(classification_report(y_test, y_pred_rf,
      target_names=['Low Risk', 'High Risk']))

print(f"\nConfusion Matrix:")
cm = confusion_matrix(y_test, y_pred_rf)
print(f"{'':15s} {'Pred Low':>10s} {'Pred High':>10s}")
print(f"{'Actual Low':15s} {cm[0,0]:>10d} {cm[0,1]:>10d}")
print(f"{'Actual High':15s} {cm[1,0]:>10d} {cm[1,1]:>10d}")
```

# Calibrate the model:

```python
from sklearn.calibration import CalibratedClassifierCV
import joblib

best_rf = joblib.load('burnout_model.pkl')

calibrated_model = CalibratedClassifierCV(best_rf, method='isotonic', cv=5)
calibrated_model.fit(X_train, y_train)

print("Before calibration:")
probs_before = best_rf.predict_proba(X_test)[:, 1]
print(f"  Min: {probs_before.min():.3f}  Max: {probs_before.max():.3f}  Mean: {probs_before.mean():.3f}")

print("After calibration:")
probs_after = calibrated_model.predict_proba(X_test)[:, 1]
print(f"  Min: {probs_after.min():.3f}  Max: {probs_after.max():.3f}  Mean: {probs_after.mean():.3f}")
```

## 🔹 Interpretation of Probability Calibration Results

### 📊 Results

**Before calibration:**
- Min: 0.027  
- Max: 0.983  
- Mean: 0.514  

**After calibration:**
- Min: 0.002  
- Max: 1.000  
- Mean: 0.514  

---

### 🧠 Analysis

1. **Wider probability range after calibration**
   - The minimum probability decreased (0.027 → 0.002)
   - The maximum probability increased (0.983 → 1.000)

   👉 This indicates that the calibrated model can express **stronger certainty** in both low-risk and high-risk predictions.

---

2. **Unchanged mean probability**
   - The mean remains **0.514 before and after calibration**
   - This shows that calibration does **not change the overall distribution balance**
   - It only reshapes how probabilities are distributed across samples

---

3. **Effect of isotonic calibration**
   - Adjusts raw model outputs into more realistic probability estimates
   - Preserves ranking of predictions
   - Improves probability reliability without changing model accuracy

---

### 🎯 Conclusion

Probability calibration improves the **interpretability and reliability** of the model’s predicted risk scores by refining probability estimates, while maintaining the same overall prediction distribution.
